# Data manipulation

The following contains common data operations that we'll perform when analyzing data within a dataframe

```python
data = pd.read_csv("../data/gapminder_gdp_europe.csv", index_col="country")
```

```python
print(data)
```

```python
data.head()
```

We can access of query specific rows or columns by index or by name

```python
print(data.iloc[0, 0])
```

```python
print(data.loc["Albania", "gdpPercap_1952"])
```

Sometimes we want to collect all the values in a row, or in a column, we do that using the special character':' to tell padas to return all values

```python
print(data.loc["Albania", :])
```

```python
print(data.loc[:, "gdpPercap_1952"])
```

Sometimes we want to collect multiple rows and/or columns all at once

```python
print(data.loc["Italy":"Poland", "gdpPercap_1962":"gdpPercap_1972"]) 
```

We can do operations and calculations on the "view" of our dataframe without changing the dataframe itself

```python
print(data.loc["Italy":"Poland", "gdpPercap_1962":"gdpPercap_1972"].max())
```

## More complicated data queries

Many times we have complex questions for queries we want to ask of the data, how do we do that?

```python
subset = data[data > 10000]
print(subset)
```

```python
print(subset.loc["Italy":"Poland", "gdpPercap_1962":"gdpPercap_1972"])
```

```python
subset = data.loc["Italy":"Poland","gdpPercap_1962":"gdpPercap_1972"]
```

```python
print("Where are GDP higher than 10K in our subset?\n", subset > 10000)
```

We can use the results of our query for further investigation without changing our dataframe

```python
print(subset[subset > 10000].describe())
```

## Split, apply, combine operations on dataframes

We can also perform more complicated manipulation of data in dataframes beyond the builtin functions and variables that the dataframe contains

```python
mask_higher = data > data.mean()
wealth_score = mask_higher.aggregate("sum", axis=1) / len(data.columns)
print(wealth_score)
```

```python
data.mean(axis=0)
```
